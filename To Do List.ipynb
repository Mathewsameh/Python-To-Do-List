```python
print("=" * 5 + "To Do List" + "=" * 5)

print("1. Add task")
print("2. Show task")
print("3. Delete task")
print("4. Exit")
print("\n")
theChoice = input("Please choose from (1 to 4): ").strip()

while not theChoice.isnumeric() or int(theChoice) < 1 or int(theChoice) > 4:
    print("Error, please try again")
    theChoice = input("Please choose from 1 to 4: ").strip()

theChoice = int(theChoice)

print("=" * 50)
print("Your number is successfully")
# ===============================
# Choice 1 => Add task
# ===============================

if theChoice == 1:
    myFile = open("tasks.txt", "a")
    
    while True:
        theTask = input('Enter your task (or type "done" to finish): ').strip().capitalize()
        
        if theTask == "Done":
            break
            
        myFile.write(theTask + "\n")
    myFile.close()
# ===============================
# Choice 2 => Show task
# ===============================
elif theChoice == 2:
    myFile = open("tasks.txt", "r")
    tasks = myFile.read()
    myFile.close()
    
    if len(tasks) == 0:
        print("No tasks yet")
        
    if len(tasks) > 0:
        print("=" * 20)
        print("Your tasks are:")
        print("=" * 20)
        print(tasks)
# ===============================
# Choice 3 => Delete task
# ===============================
elif theChoice == 3:
    myFile = open("tasks.txt", "r")
    tasks = myFile.readlines()
    myFile.close()
    
    if len(tasks) == 0:
        print("No tasks yet")
        
    if len(tasks) > 0:
        print("=" * 20)
        print("Your tasks are:")
        print("=" * 20)
        
        for taskNumber, task in enumerate(tasks, start=1):
            print(f"{taskNumber}. {task.strip()}")
        print("=" * 20)

        theNewChoice = input("Please enter the task number to delete: ").strip()

        while not theNewChoice.isnumeric() or int(theNewChoice) < 1 or int(theNewChoice) > len(tasks):
            print("Error, please enter a valid task number")
            theNewChoice = input("Please enter the task number to delete: ").strip()
            
        theNewChoice = int(theNewChoice)
        tasks.pop(theNewChoice - 1)
        
        myFile = open("tasks.txt", "w")
        
        for task in tasks:
            myFile.write(task)
        myFile.close()

        print("Task deleted successfully")
# ===============================
# Choice 4 => Exit
# ===============================
elif theChoice == 4:
    print("Goodbye!")
```
